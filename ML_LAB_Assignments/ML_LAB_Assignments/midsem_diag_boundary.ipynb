```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.inspection import DecisionBoundaryDisplay

np.random.seed(42)
X = np.random.rand(20000, 2) * 10
y = (X[:, 1] > X[:, 0]).astype(int)
```

```python
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X, y)
```

```python

plt.figure(figsize=(8, 8))

DecisionBoundaryDisplay.from_estimator(
    clf, X, response_method="predict", cmap="bwr", alpha=0.4, ax=plt.gca(), grid_resolution=500
)

# We only plot 1000 points so the graph isn't a messy blob of ink,
# but the tree learned from all 20,000!
X_plot = X[:500]
y_plot = y[:500]

plt.scatter(X_plot[y_plot==0, 0], X_plot[y_plot==0, 1], c='blue', edgecolor='k', s=20)
plt.scatter(X_plot[y_plot==1, 0], X_plot[y_plot==1, 1], c='red', edgecolor='k', s=20)

plt.xlim(0, 10)
plt.ylim(0, 10)
plt.xlabel("X_1")
plt.ylabel("X_2")
plt.title("Decision Tree Approximating a Diagonal Line")
plt.show()
```
